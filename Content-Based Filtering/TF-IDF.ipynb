## **TF-IDF (Term Frequency-Inverse Document Frequency)**


**Term Frequency-Inverse Document Frequency (TF-IDF)** — это один из наиболее распространенных и мощных методов для извлечения признаков из текстовых данных.TF-IDF вычисляет важность каждого слова в документе относительно количества его употреблений в данном документе и во всей коллекции текстов. Этот метод позволяет выделить ключевые слова и понять, какие слова имеют больший вес для определенного документа в контексте всей коллекции.

**TF (Частота термина)** обозначает, насколько часто определенное слово появляется в данном документе. Таким образом, TF измеряет важность слова в контексте отдельного документа.

**IDF (Обратная частота документа)** измеряет, насколько уникально слово является по всей коллекции документов. Слова, которые появляются в большинстве документов, имеют низкое IDF, так как они не вносят большой информационной ценности.

### **Формула вычисления TF-IDF**

Формула TF-IDF комбинирует понятия TF и IDF, чтобы вычислить важность каждого слова в каждом документе. Формально, формула выглядит следующим образом:

$$ TF-IDF(t, d) = TF(t, d) * IDF(t) $$

где:

- **TF(t, d)** - Частота термина (TF) для слова "t" в документе "d".

- **IDF(t)** - Обратная частота документа (IDF) для слова "t".

#### **Формулы для TF и IDF**

$$ TF = \frac{t}{d} $$

где:

- **t** - количество вхождений слова в документ

- **d** - общее количество слов в документе

$$ IDF = log(\frac{a}{b})$$

- **a** - общее количество документов

- **b** - количество документов, содержащих слово

Для реализации этого метода мне потребуется текст статей. Разберем этот метод на 5 статей

```python
import math
from collections import Counter
```

```python
news = [
    "Apple announces new iPhone with advanced camera features.",
    "Tesla unveils new electric car model with longer range.",
    "NASA discovers new exoplanet in distant galaxy.",
    "Google releases updates to its AI language models.",
    "SpaceX successfully launches satellite into orbit."
]

user_read_news = ["Tesla unveils new electric car model with longer range.", "Google releases updates to its AI language models."]
```

Разобьем новости на слова, тем самым получим токены

```python
def tokenize(text):
    return text.lower().split()
tokenize_corpus = [tokenize(doc) for doc in news]

print(tokenize_corpus)
```

Мы получили набор токенов, теперь с их помощью можно посчитать TF и IDF по формулам приведенными выше

```python
def compute_tf(doc):
    word_counts = Counter(doc)
    total_words = len(doc)
    tf = {word: count / total_words for word, count in word_counts.items()}
    return tf

tf_values = [compute_tf(doc) for doc in tokenize_corpus]

print(tf_values)
```

```python
def compute_idf(corpus):
    num_docs = len(corpus)
    idf = {}
    all_words = set(word for doc in corpus for word in doc)

    for word in all_words:
        doc_count = sum(1 for doc in corpus if word in doc) #считаю сколько документов содержат слово
        idf[word] = math.log(num_docs/(1+doc_count))
    
    return idf
idf_values = (compute_idf(tokenize_corpus))
print(idf_values)
```

Рассчитываем TF-IDF

```python
def comput_tf_idf(corpus):
    idf = compute_idf(corpus)
    tf_idf_matrix = []
    for doc in corpus:
        tf = compute_tf(doc)
        tf_idf = {}
        for word in tf:
            tf_idf[word] = tf[word] * idf.get(word, 0)
        tf_idf_matrix.append(tf_idf)
    return tf_idf_matrix

tf_idf_matrix = comput_tf_idf(news)
```

**Метод главных компонентов (Principal Component Analysis, PCA)** — это инструмент для снижения размерности данных и выявления скрытых паттернов.

```python
#для удобства построим матрицу признаков
def build_feature_matrix(tf_idf_matrix):
    vocabulary =set()
    for doc in tf_idf_matrix:
        vocabulary.update(doc.keys())
    vocabulary = sorted(vocabulary)
    feature_matrix = []
    for doc in tf_idf_matrix:
        row = [doc.get(word, 0) for word in vocabulary]
        feature_matrix.append(row)
    return feature_matrix, vocabulary

feature_matrix, vocabulary = build_feature_matrix( tf_idf_matrix)
```

Тут мы будем реализовывать метод главных компонентов:

- Центрируем данные

- Вычисляем ковариантную матрицу

- Находим собственные значения и векторы

- Проектируем данные на главные компоненты

```python
def mean_center(matrix):
    means = [sum(col)/len(col) for col in zip(*matrix)] #транспонируем матрицу и вычисляем для каждого столбца среднее значение

    #Центрирую данные вычитая среднее значение столбца
    centered_matrix = [
        [row[i]-means[i] for i in range(len(row))] #Для каждого элемента строки
        for row in matrix #Перемещение по строкам
        ]
    return centered_matrix

#Создаем ковариационную матрицу
def covariance_matrix(matrix):
    n = len(matrix)

    #Создаем пустую ковриационную матрицу
    cov_matrix = [
        [0]*len(matrix[0])
        for _ in range(len(matrix[0]))
    ]

    #Заполняем её
    for i in range(len(matrix[0])):
        for j in range(len(matrix[0])):
            cov_matrix[i][j]=sum(matrix[k][i] * matrix[k][j] for k in range(n)) / (n-1)
    return cov_matrix

#Найдем собственные значения и собственные векторы
def eigen(matrix):
    from numpy.linalg import eig
    eigenvalues, eigenvectors = eig(matrix)
    return eigenvalues, eigenvectors

#Реализуем PCA
def pca(matrix, n_components):
    centered_matrix = mean_center(matrix)
    cov_matrix = covariance_matrix(centered_matrix)
    eigenvalues, eigenvectors = eigen(cov_matrix)

    idx = sorted(range(len(eigenvalues)), key=lambda i: eigenvalues[i], reverse=True) #Сортируем индексы по убыванию их величины
    eigenvectors = eigenvectors[:, idx[:n_components]]

    reduced_matrix = [
        [sum(row[j]*eigenvectors[j][i] for j in range(len(row)))
         for i in range(n_components)]
         for row in centered_matrix
    ]
    return reduced_matrix

reduced_matrix = pca(feature_matrix, n_components=2)
```

```python
def cosine_similarity(vec1, vec2):
    dot_product = sum(a * b for a, b in zip(vec1, vec2))
    magnitude1 = math.sqrt(sum(a ** 2 for a in vec1))
    magnitude2 = math.sqrt(sum(a ** 2 for a in vec1))
    return dot_product / (magnitude1 * magnitude2)

def recommend_news(user_read_indices, reduced_matrix, news):
    #Создаем профиль пользователя
    user_profile = [reduced_matrix[i] for i in user_read_indices]
    user_profile = [sum(x) / len(user_profile) for x in zip(*user_profile)]

    #Вычисляем косинусное подобие
    similarities = []
    for i, vec in enumerate(reduced_matrix):
        similarity = cosine_similarity(user_profile, vec)
        similarities.append((i, similarity))

    #Сортируем по подобию
    recommendations = sorted(similarities, key=lambda x: x[1], reverse=True)
    
    return [news[i] for i, _ in recommendations if i not in user_read_indices]

user_read_indices = [news.index(doc) for doc in user_read_news]

recommended_news = recommend_news(user_read_indices, reduced_matrix, news)
print("Рекомендуемые новости: ")
for news_item in recommended_news:
    print("-", news_item)
```
