### Метод TF-IDF

Я не смог разобрать таблицу news поэтом упока, просто проверю работу этого метода

Для реализации этого метода мне потребуются

```python
import math
from collections import Counter
```

```python
news = {
    "Apple announces new iPhone with advanced camera features.",
    "Tesla unveils new electric car model with longer range.",
    "NASA discovers new exoplanet in distant galaxy.",
    "Google releases updates to its AI language models.",
    "SpaceX successfully launches satellite into orbit."
}
```

Разобьем новости на слова

```python
def tokenize(text):
    return text.lower().split()
tokenize_corpus = [tokenize(doc) for doc in news]

print(tokenize_corpus)
```

Мы получили набор токенов, теперь с помощью этих токенов можно посчитать TF и IDF по формулам (потом нормально оформлю)

TF = кол-во вхождений слова в документ/общ кол-во слов в документе

IDF = log(общ кол-во документов/кол-во документов, содержащих слово)

```python
def compute_tf(doc):
    word_counts = Counter(doc)
    total_words = len(doc)
    tf = {word: count / total_words for word, count in word_counts.items()}
    return tf
tf_values = [compute_tf(doc) for doc in tokenize_corpus]

print(tf_values)
```

```python
def compute_idf(corpus):
    num_docs = len(corpus)
    idf = {}
    all_words = set(word for doc in corpus for word in doc)

    for word in all_words:
        doc_count = sum(1 for doc in corpus if word in doc) #считаю сколько документов содержат слово
        idf[word] = math.log(num_docs/(1+doc_count))
    
    return idf
idf_values = (compute_idf(tokenize_corpus))
print(idf_values)
```

TF-IDF считается по вормуле: TF * IDF

```python
def comput_tf_idf(tf, idf):
    tf_idf = {}
    for word, tf_value in tf.items():
        tf_idf[word] = tf_value*idf.get(word,0)
    return tf_idf
tf_idf_values = [comput_tf_idf(tf, idf_values) for tf in tf_values]

print("TF-IDF значения:")
for i, tf_idf in enumerate(tf_idf_values):
    print(f"Новость{i+1}:")
    for word, score in tf_idf.items():
        print(f"{word}: {score:.4f}")
```
